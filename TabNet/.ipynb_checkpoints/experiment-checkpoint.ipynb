```python
import pandas as pd
import numpy as np

import tensorflow as tf
from tensorflow.keras import layers, models, optimizers
from tensorflow.keras.losses import BinaryCrossentropy

from sklearn.preprocessing import LabelEncoder
```

```python
data = pd.read_csv("Final.csv")

data = data[["Crop", "Stage", "Disease Name"]]
data.head()
```

```python
grouped = (
    data
    .groupby(["Crop", "Stage"])["Disease Name"]
    .apply(list)
    .reset_index()
)

grouped.head()
```

```python
all_diseases = sorted(
    data["Disease Name"].unique().tolist()
)

disease_to_index = {
    d: i for i, d in enumerate(all_diseases)
}

num_diseases = len(all_diseases)
num_diseases
```

```python
Y = np.zeros((len(grouped), num_diseases))

for row_idx, diseases in enumerate(grouped["Disease Name"]):
    for d in diseases:
        Y[row_idx, disease_to_index[d]] = 1

Y.shape
```

```python
crop_le = LabelEncoder()
stage_le = LabelEncoder()

grouped["Crop_enc"] = crop_le.fit_transform(grouped["Crop"])
grouped["Stage_enc"] = stage_le.fit_transform(grouped["Stage"])

X = grouped[["Crop_enc", "Stage_enc"]].values

X.shape
```

```python
def build_tabnet(input_dim, output_dim):
    inputs = layers.Input(shape=(input_dim,))

    x = layers.Dense(64, activation="relu")(inputs)
    x = layers.BatchNormalization()(x)

    x = layers.Dense(64, activation="relu")(x)
    x = layers.BatchNormalization()(x)

    outputs = layers.Dense(
        output_dim,
        activation="sigmoid"   # 🔥 REQUIRED for multi-label
    )(x)

    model = models.Model(inputs, outputs)
    return model
```

```python
model = build_tabnet(
    input_dim=X.shape[1],
    output_dim=Y.shape[1]
)

model.compile(
    optimizer=optimizers.Adam(learning_rate=0.001),
    loss=BinaryCrossentropy()
)

model.summary()
```

```python
history = model.fit(
    X,
    Y,
    epochs=200,
    batch_size=8,
    verbose=1
)
```

```python
def predict_diseases(crop_name, stage_name, threshold=0.4):
    crop_enc = crop_le.transform([crop_name])[0]
    stage_enc = stage_le.transform([stage_name])[0]

    x_input = np.array([[crop_enc, stage_enc]])

    probs = model.predict(x_input)[0]

    predicted = [
        all_diseases[i]
        for i, p in enumerate(probs)
        if p >= threshold
    ]

    return predicted
```

```python
for i in range(len(grouped)):
    crop = grouped.iloc[i]["Crop"]
    stage = grouped.iloc[i]["Stage"]

    predicted = set(predict_diseases(crop, stage, threshold=0.4))
    actual = set(grouped.iloc[i]["Disease Name"])

    if predicted != actual:
        print("MISMATCH")
        print("Crop:", crop)
        print("Stage:", stage)
        print("Actual:", actual)
        print("Predicted:", predicted)
        break
else:
    print(" PERFECT MATCH FOR ALL ROWS")
```
